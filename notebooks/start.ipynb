# Start here!

Welcome to the **Glashelder** notebooks. This is an outline of the `notebooks/` directory — use it to figure out which notebook fits the task you want to experiment with.

Each subdirectory here corresponds to a workspace package in `src/` and demonstrates that package's functionality interactively.

For background on the repository as a whole, see the top-level [README](../README.md).

## Pipeline overview

A typical end-to-end flow through the experiments looks like this:

```mermaid
flowchart LR
  raw[Raw glass plate scans] --> pre[preprocessing<br/>YOLO crop]
  pre --> emb[embed<br/>CLIP/DinoV2]
  emb -->|CLIP/DinoV2| clu[cluster<br/>UMAP + HDBSCAN]
  emb -->|CLIP| srx[search<br/>cosine similarity]
  pre --> art[artmatcher<br/>match known artworks]
```

You don't have to run them in order — each notebook is self-contained. Start with whichever experiment you're interested in.

## Notebooks

### [preprocessing/yolo.ipynb](./preprocessing/yolo.ipynb)

YOLOv11-based preprocessing that crops the photographic content out of a glass plate scan (removing borders and description text). Covers:
- Running inference with the pretrained Glashelder `best.pt` model on a folder of images
- Creating your own YOLO dataset (annotation in LabelStudio) and training a new model

**Start here if:** you have raw scans and want cropped photographic content for downstream tasks.

### [cluster/cluster.ipynb](./cluster/cluster.ipynb)

Cluster the Glashelder dataset using CLIP embeddings reduced with UMAP and clustered with HDBSCAN. A second UMAP pass reduces to 3D for visualization (Rerun).

**Start here if:** you want to discover groupings in an unlabeled image set.

### [cluster/search.ipynb](./cluster/search.ipynb)

Text-to-image search interface built on CLIP embeddings. Encodes a text prompt into the same latent space as the images and serves the best matches through a local Flask web UI at `localhost:5000`.

**Start here if:** you want to query an image collection with natural language.

### [classification/classify.ipynb](./classification/classify.ipynb)

CLIP-based multi-label classification using user-supplied prompts. Generates embeddings and assigns each image to the best-matching label (or `not_classified` if no label clears the threshold).

**Start here if:** you have target categories in mind and want to tag images against them.

### [artmatcher/artmatcher.ipynb](./artmatcher/artmatcher.ipynb)

Random Forest classifier that decides whether a query image genuinely matches a reference artwork. Combines 23 features across three categories: embedding-similarity distribution stats, ORB keypoint geometric verification, and perceptual hashes (pHash/dHash/wHash/aHash).

**Start here if:** you have a set of known artworks and want to find glass plates depicting them.

## Running the notebooks

You essentially have two options for running the notebooks:

1. using the Jupyter Lab web UI
2. within Visual Studio Code, using the `uv` managed python interpreter

### 1. Using the Jupyter Lab web UI

Launch Jupyter through `uv run`:

```bash
uv run jupyter lab
```

### 2. Using Visual Studio Code

Open the notebook files in VSC. When prompted to select an interpreter make sure to use `.venv/bin/python` on Linux/MacOS and `.venv\scripts\python` on Windows.

Most notebooks also expose their logic as CLI commands — see `uv run glashelder --help` (more information [here](../README.md#cli)).

## A note on paths

Some notebooks use paths relative to the notebook file's location (`./output`, `../../README.md`, etc.). Those resolve correctly when the Jupyter kernel runs locally in the notebook's directory. If you attach a **remote/external kernel**, the kernel's working directory won't match the notebook's location and relative paths will break — in that case, anchor paths to the git repo root or set them explicitly at the top of the notebook.

You should therefore use **absolute paths** wherever you can.
